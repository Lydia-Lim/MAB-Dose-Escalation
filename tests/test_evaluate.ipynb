### Test DLE Plot

```python
def generate_n_dles():
    n_dles = list(map(int, list(
        np.round(np.random.normal(
            np.random.choice(27), 
            np.random.choice(range(1, 4)), 
            size=1000
        ))
    )))
    return [
        n_dle if n_dle >= 0 else 0 
        for n_dle in n_dles
    ]
```

```python
def generate_algo_n_dle_map(algos):
    return {algo: generate_n_dles() for algo in algos}
```

```python
ALGOS = ["3 + 3", "CRM", "DP-SL"]
N_DLE_MAP = {
    "a = 0.4": generate_algo_n_dle_map(ALGOS), 
    "a = 1": generate_algo_n_dle_map(ALGOS), 
    "a = 1.3": generate_algo_n_dle_map(ALGOS), 
    "a = 3.4": generate_algo_n_dle_map(ALGOS), 
}
```

```python
plot_n_dles(N_DLE_MAP, show_fig=True)
```
